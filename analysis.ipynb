```python
import csv

# Opens the Food dataset so we can read its contents
food_file = open("food.csv", "r", newline="")

# Uses csv.reader to correctly separate values into lists
food_data = list(csv.reader(food_file))

food_file.close()

# Checks the column names and first food entry
print(food_data[0])
print(food_data[1])
```

```python
# Finds the correct columns for comparing protein and fat
protein_index = food_data[0].index("Data.Protein")
fat_index = food_data[0].index("Data.Fat.Total Lipid")

count = 0

print("Food | Protein (g) | Fat (g)")
print("-" * 65)

# Limits the results to 10 high-protein foods for easier comparison
for food in food_data[1:]:
    if food[protein_index] != "" and float(food[protein_index]) > 20:
        print(food[1][:35], "|", food[protein_index], "|", food[fat_index])
        count += 1

        if count == 10:
            break
```
